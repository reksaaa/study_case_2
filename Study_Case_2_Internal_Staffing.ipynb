```python
from sklearn.linear_model import LinearRegression
import numpy as np


def create_sequence_features(series, window_size=3):
    """
    Mengubah data 1D menjadi fitur X dan target y
    menggunakan metode Sliding Window.
    """

    X, y = [], []

    for i in range(len(series) - window_size):
        # Mengambil 3 data sebelumnya sebagai fitur
        X.append(series[i:i + window_size])

        # Mengambil data berikutnya sebagai target
        y.append(series[i + window_size])

    return np.array(X), np.array(y)


def train_and_predict_next(train_data, test_data_window):
    """
    Melatih model Linear Regression dan memprediksi
    satu titik waktu berikutnya.
    """

    # Ekstraksi fitur dari data training
    X_train, y_train = create_sequence_features(
        train_data,
        window_size=3
    )

    # Inisialisasi model
    model = LinearRegression()

    # Melatih model
    model.fit(X_train, y_train)

    # Mengubah test window menjadi bentuk 2D: (1, 3)
    test_data_window = np.array(test_data_window).reshape(1, -1)

    # Melakukan prediksi
    prediction = model.predict(test_data_window)[0]

    # Memastikan hasil berupa float
    return float(prediction)


# Uji coba fungsi
predicted_value = train_and_predict_next(
    train_series,
    test_window
)

print(f"Prediksi Glukosa Selanjutnya: {predicted_value}")
print(f"Nilai Aktual Glukosa: {actual_next_value}")
```
